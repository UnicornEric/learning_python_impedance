## Concept and API reminders
Final evaluation estimates how a frozen procedure performs under its declared patient population and target. It is valid only insofar as selection did not adapt to those test outcomes. Reporting prespecified errors/subgroups describes the fixed model; modifying it afterward starts a new development cycle.

A patient-bootstrap interval around the paired MAE difference expresses conditional sampling variability, not all possible model-development uncertainty. Statistics and ML contribute complementary evidence: an association describes a model-adjusted relationship; prediction measures useful generalization under the split; neither alone establishes causal or clinical utility.

The handoff should let a colleague understand what was measured, what was learned, what was independently evaluated, how to reproduce inference, and which evidence is still missing.
